```python

import gradio as gr
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Frequently asked questions and answers
faqs = {
    "What is artificial intelligence?":
        "Artificial Intelligence (AI) enables machines to perform tasks that normally require human intelligence.",

    "What is machine learning?":
        "Machine Learning is a branch of AI that allows computers to learn patterns from data.",

    "What is Python?":
        "Python is a beginner-friendly programming language used in AI, web development, and data analysis.",

    "What is an internship?":
        "An internship provides practical work experience and helps students develop professional skills.",

    "What is CodeAlpha?":
        "CodeAlpha provides internship opportunities and practical project experience for students.",

    "How can I contact support?":
        "Please check the official website of the organization for its support contact details.",

    "Thank you":
        "You're welcome! Feel free to ask another question.",

    "What are your working hours?":
        "Please check the relevant organization's official website for its working hours."
}

questions = list(faqs.keys())

# Convert questions into numerical text features
vectorizer = TfidfVectorizer()
question_vectors = vectorizer.fit_transform(questions)

def chatbot(user_question):
    if not user_question or not user_question.strip():
        return "Please type a question."

    user_vector = vectorizer.transform([user_question])
    similarities = cosine_similarity(
        user_vector, question_vectors
    )[0]

    best_index = similarities.argmax()
    best_score = similarities[best_index]

    if best_score < 0.15:
        return "Sorry, I don't know the answer. Please try another question."

    best_question = questions[best_index]
    return faqs[best_question]

app = gr.Interface(
    fn=chatbot,
    inputs=gr.Textbox(
        label="Ask your question",
        placeholder="Example: What is Python?"
    ),
    outputs=gr.Textbox(label="Chatbot answer"),
    title="FAQ Chatbot",
    description="Ask questions about AI, Python, internships, and CodeAlpha."
)

app.launch(share=True)
```
